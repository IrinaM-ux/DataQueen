Aquest exercici se centra en crear una funció que ajudi a crear (i exportar) un gràfic automàticament per a cada columna del dataframe. Per exemple, un histograma/boxplot si la variable és numèrica, unes barres dels valors més freqüents si és categòrica i unes barres dels anys més freqüents si la dada està en format data.

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

def explorar_dataset(df):
    for columna in df.columns:
        plt.figure(figsize=(6, 4))

        if pd.api.types.is_numeric_dtype(df[columna]):
            sns.histplot(df[columna].dropna(), kde=True)
            plt.title(f"Histograma: {columna}")

        elif pd.api.types.is_datetime64_any_dtype(df[columna]):
            anys = df[columna].dt.year.value_counts().sort_index()
            sns.barplot(x=anys.index, y=anys.values)
            plt.title(f"Anys més freqüents: {columna}")
            plt.xticks(rotation=45)

        else:
            top_valors = df[columna].value_counts().head(10)
            sns.barplot(x=top_valors.values, y=top_valors.index)
            plt.title(f"Valors més freqüents: {columna}")

        plt.tight_layout()
        plt.savefig(f"grafic_{columna}.png")
        plt.show()
```

```python
titanic = sns.load_dataset("titanic")
explorar_dataset(titanic)
```

## Exercici 2: Optimització de rutes (problema aplicat)

🌍 Context: L'empresa no disposa d'un control fiable del quilometratge dels desplaçaments dels seus treballadors. Abans d'implementar un sistema avançat, es vol fer una prova pilot amb treballadors ubicats a Espanya.
Per a aquesta prova:

» Es treballa amb una matriu de distàncies entre ciutats.
»Es vol una solució simple, raonable i explicable.
» No es busca la ruta òptima, sinó una aproximació coherent.
Carrega l'arxiu matriu_distancies.xlsx a pandas de manera que les files i columnes siguin les ciutats i s'eliminin "Las Palmas de Gran Canaria" i "Palma".

Desenvolupa una funció que:

» Rebi: una ciutat d'origen i una ciutat de destinació.
» Retorni: l'ordre de les ciutats visitades (ruta) i la distància total recorreguda.
La ruta s'ha de construir seguint una estratègia senzilla: anar sempre a la ciutat no visitada més propera fins a arribar a la destinació.

```python
matriu_distancies = pd.read_excel(r"/content/matriu_distancies.xlsx", index_col=0)
matriu_distancies = matriu_distancies.drop(columns=["Las Palmas de Gran Canaria", "Palma"])
display(matriu_distancies)
```

- he construido la funcción guiandome por las distancias entre ciudades
- he utilizado el bucle while porque la ciudad de referencia cambia cada vez. La distancia se calcula cada vez desde la ciudad actual, no desde la ciudad de origen
- he definido la siguiente ciudad de la ruta con la funcción .idxmin() que indica la distancia más corta en la serie, excluyendo las ciudades ya visitadas con .drop()
- en paralel a la variable ruta que recoge las ciudades visitadas, he sumado las distancias correspondientes en la variable distancia

```python
def busca_ciutat(ciutat_origen, ciutat_desti):
    ruta = [ciutat_origen]
    ciutat_actual = ciutat_origen
    distancia = 0

    while ciutat_actual != ciutat_desti:
      seguent_ciutat = matriu_distancies.loc[ciutat_actual].drop(ruta).idxmin()
      distancia += matriu_distancies.loc[ciutat_actual, seguent_ciutat]
      ruta.append(seguent_ciutat)
      ciutat_actual = seguent_ciutat
    return ruta, float(distancia)

busqueda = busca_ciutat("Barcelona", "Murcia")

print(busqueda)
```
