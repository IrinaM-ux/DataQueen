### 1. obtenir_dades()
- la funcción crea un diccionario para almacenar los datos para cada candidato;
- pide los datos requeridos a cada candidato;
- he transformado el input para antigüedad, catalan, Excel y titulo academico en integer (porque los necesitaré para calculos)
- la funcción devuelve un diccionario por candidato.

```python
def obtenir_dades():

  candidat = {}
  candidat["nom"] = input("Nom i cognoms?")
  candidat["telefon"] = input("Telèfon de contacte?")
  candidat["antiguedat"] = int(input("Antigüedat?"))
  candidat["catala"] = int(input("Català?"))
  candidat["Excel"] = int(input("Excel?"))
  candidat["titol_academic"] = int(input("Títol acadèmic?"))
  return candidat

```

```python
obtenir_dades()
```

### 2. validar_dades()
- la funcción define los valores posibles/longitud exacta para cada variable y establece que la casilla del nombre no puede quedarse en blanco;
- devuelve las variables validadas


```python

def validar_dades(candidat):

  catala_valid = candidat["catala"] in [0, 1, 2, 3]
  excel_valid = candidat["Excel"] in [0, 1, 2, 3]
  titol_valid = candidat["titol_academic"] in [0, 1, 2, 3]
  telefon_valid = len(candidat["telefon"])==9
  antiguedat_valid = candidat["antiguedat"]>=0
  nom_valid = candidat["nom"].strip()!= ""

  return catala_valid and excel_valid and titol_valid and telefon_valid and antiguedat_valid and nom_valid

```

### 3. calcular_puntuacio()
- la funcción añade el elemento "puntuación" al diccionario;
- define la formula para calcular el valor de la puntuación para cada candidato;
- devuelve el diccionario actualizado con datos de un candidato;

```python
def calcular_puntuacio(candidat):
  candidat ["puntuacio"] = candidat["antiguedat"]*1.5 + candidat["catala"] + candidat["Excel"] + candidat["titol_academic"]
  return candidat
```

### 4. classificar_candidat()
- la funcción evalua el diccionario con los datos del candidato
- crea un elemento nuevo "classificació" en el diccionario
- clasifica cada candidato en funcción de su puntuación.
- devuelve el diccionario actualizado.


```python
def classificar_candidat(candidat):
  if candidat ["puntuacio"] >= 12:
    candidat["classificacio"] = "Prioritat Alta"
  elif candidat ["puntuacio"] >=8:
    candidat["classificacio"] = "Candidat adequat"
  else:
    candidat["classificacio"] = "No prioritari"
  return candidat
```

```python
def mostrar_individual(candidat):
  print(f"{candidat['nom']}:{candidat['puntuacio']}")
```

### 6. mostrar_informe()
- la funcción recibe como argumento un diccionario con todos los candidatos.
- establece primero las reglas para ordenar los candidatos en funcción de los valores de clasificación
- despues ordena los candidatos según las prioridades establecidas y según puntuación.
- almacena el nombre y el teléfono de los candidatos ordenados por clasificacion y puntuación en la variable "ranking"
- imprime uno por uno los elementos de ranking.



```python
def mostrar_informe(candidats):
    ordre_prioritat = {"Prioritat Alta": 3, "Candidat adequat": 2, "No prioritari": 1}
    ranking = sorted(candidats, key=lambda c: (ordre_prioritat[c['classificacio']], c['puntuacio']), reverse=True)

    print("Informe final:")
    for c in ranking:
        print(f"{c['nom']} - {c['telefon']} - {c['puntuacio']}")
```

### 1. obtener_segmento_y_factor()
- recibe un codi_comportament

### 4. mostrar_diccionari()

- devuelve un diccionario nuevo, donde se han añadido las infos nuevas en cada ficha de cliente
- imprime el diccionario resultats

```python
def processar_clients(clients):
  resultats = {}
  for cliente in clients:
     codi = cliente["codi_comportament"]
     grup_edat = cliente["grup_edat"]
     factor_codi, segment, missatge = obtenir_segment_i_factor(codi)
     factor_edat_val = obtenir_factor_edat(grup_edat)
     risc_final = calcular_risc(factor_codi, factor_edat_val)
     resultats[cliente["id_client"]] = {"id_cliente": cliente["id_client"], "nombre": cliente["nom"], "edad": cliente["edat"], "grupo_edad": cliente["grup_edat"], "codigo_comportamiento": cliente["codi_comportament"], "segmento": segment, "mensaje": missatge, "factor_codigo": factor_codi, "factor_edad": factor_edat_val, "riesgo_final": risc_final}
  return resultats

processar_clients(clients)
```

## **Exercici 1: Gestió de risc i priorització de pacients**

```python
pacients = [
    "Maria|45|tos, Fiebre Alta, dolor_pecho",
    "Luis|33|dolor_cabeza,faTiga",
    "Sara|67|fiebre alta, dificultad RESPIRAR, tos",
    "Jordi|52|fatiga , tos",
    "Anna|29|fiebre alta,dolor_cabeza"
]
```

```python
nivells_risc = {
    "fiebre_alta": 3,
    "dificultad_respirar": 5,
    "dolor_pecho": 4,
    "tos": 1,
    "fatiga": 1,
    "dolor_cabeza": 1
}
```

### 1. separar_info_pacient()
-la funcción recibe la ficha / el diccionario de un paciente
- utiliza el método split para "romper" el texto donde aparece el simbolo |
- almacena los "trozos" de texto en la variable "parts"
- transforma en integer el primer elemento de la lista parts
- y lo almacena en la variable edat
- devuelve un diccionario con el nombre, edad y simptomas del paciente


```python
def separar_info_pacient(text):
  parts = text.split("|")
  edat = int(parts[1])

  return {"nom":parts[0], "edat":edat, "simptomes": parts[2]}
```

```python
separar_info_pacient("Luis|33|dolor_cabeza,faTiga")
```

### 2. neteja_simptomes()
- recibe como argumento el valor de "simptomes"
- rompe el texto donde hay una coma
- almacena los trozos resultantes en la variable llista
- crea la lista llista_neta
- procesa, uno por uno, cada texto sobre simptomas
- le quita los espacios (strip), pone todo en mínusculas (lower) y reemplaza comas por guiones bajos
- almacena los resultados de la limpieza en llista_neta
- devuelve llista_neta

```python
def neteja_simptomes(text_simptomes):
  llista = text_simptomes.split(",")
  llista_neta = []
  for s in llista:
    simptoma_net = s.strip().lower().replace(" ", "_")
    llista_neta.append(simptoma_net)
  return llista_neta
```

```python
print(neteja_simptomes('fiebre alta, dificultad RESPIRAR, tos'))
```

### 3. obtenir_factor_risc()
- la funcción extrae la puntuación asociada con el riesgo de cada sintoma
- y la almacena en la variable factor_risc
- devuelve factor_risc

```python
def obtenir_factor_risc(simptoma_net):
  factor_risc = nivells_risc[simptoma_net]
  return factor_risc


```

```python
obtenir_factor_risc('tos')
```

### 4. calcular_risc()
- la funcción recibe una lista de simptomas
- parte de la premisa de riesgo 0
- analiza, uno por uno, cada simptoma en llista_neta
- busca su correspondiente factor de riesgo llamando la funcción obtenir_factor_risc
- va sumando el riesgo de cada simptoma
- devuelve el valor total de riesgo de los simptomas de la lista bajo la variable risc

```python
def calcular_risc(llista_neta):
  risc = 0
  for simptoma_net in llista_neta:
    factor_risc = obtenir_factor_risc(simptoma_net)
    risc += factor_risc
  return risc
```

```python
calcular_risc(["fiebre_alta", "dificultad_respirar", "tos"])
```

### 5. procesar_paciente()
- recibe un texto sobre un paciente
- separa la info llamando la funcción ya existente
- almacena la info separada en dades
- limpia la info llamando otra funcción ya existente
- almacena la info en simptomes_nets
- calcula el riesgo llamando una tercera funcción ya existente
- almacena el riesgo en risc_total
- devuelve una ficha del paciente enriquecida con los conceptos edad, simptomes y risc_total




```python
def processar_pacient(text):
  dades = separar_info_pacient(text)
  simptomes_nets = neteja_simptomes(dades["simptomes"])
  risc_total = calcular_risc(simptomes_nets)

  return {"nom":dades["nom"], "edat":dades["edat"], "simptomes":simptomes_nets, "risc_total":risc_total}
```

```python
processar_pacient("Maria|45|tos, Fiebre Alta, dolor_pecho")
```

### 5. classificar_pacients()
- recibe un listado con pacientes procesados
- ordena los registros del listado por el valor del risc_total de mayor a menor
- devuelve el listado ordenado

```python
def classificar_pacients(pacients_processats):
  ranking_risc = sorted(pacients_processats, key=lambda info: info['risc_total'], reverse=True)
  return ranking_risc
```

### 6. mostrar_pacients()
- recibe como argumento una lista que se crea más tarde en la main
- coge cada elemento del listado ordenado
- e imprime la información de nombre, edad y riesgo total


```python
def mostrar_pacients(pacients_processats):
  ranking_risc = classificar_pacients(pacients_processats)
  for info in ranking_risc:
    print(f"{info['nom']}, {info['edat']}, {info['risc_total']}")
```

### 7. main_pacients()
- la funcción crea un diccionario para pacientes procesados
- después proces, uno por uno, cada paciente
- los añade al diccionario creado
- llama la funccción classificar_pacientes que recibe como argumento el diccionario llenado en la primera parte de la funcción main
- muestra los pacientes ordenados y clasificados
- devuelve el diccionario pacients_classificats

```python
def main_pacients():
  pacients_processats = []
  for text in pacients:
    pacients_processats.append(processar_pacient(text))
  pacients_classificats = classificar_pacients(pacients_processats)
  mostrar_pacients(pacients_processats)
  return pacients_classificats
```

```python

nivells_seguro = {
    0: {"categoria": "Baix risc", "preu_base": 120, "fraude": 0.01},
    1: {"categoria": "Risc mitjà", "preu_base": 200, "fraude": 0.05},
    2: {"categoria": "Risc alt", "preu_base": 350, "fraude": 0.15},
    3: {"categoria": "Risc crític", "preu_base": 500, "fraude": 0.30}
}
```

### 1. obtenir_dades()
- recibe como argumento un número de código de cliente
- extrae el precio base, categoria y probabilidad de fraude correspondiente a cada código

```python
def obtenir_dades(codi):
  dades = nivells_seguro[codi]
  return dades
```

```python
print(obtenir_dades(2))
```

### 2. calcular_preu_final()
- recibe como argumentos preu-base y edat
- utiliza datos de 2 diccionarios
- calcula precio final con la formula preu_final = preu_base * (1 + edat/100)
- devuleve precio final

```python
def calcular_preu_final(preu_base, edat):
  preu_final = round(preu_base * (1 + edat/100), 2)
  return preu_final
```

```python
calcular_preu_final(120, 42)
```

### 3. processar_client(text)
- recibe como agumento un texto, por ejempo una ficha de cliente en forma de diccionario
- define las variables edat, codi, preu-base como valores dentro de los diccionarions text y dades
- define la variable dades llamando la función obtenir_dades
- define la variable preu_final llamando la funcion calcular_preu_final
- devuelve un diccionario que contiene elementos de los 2 diccionarios dados y elementos calculados

```python
def processar_client(text):
  edat = text["edat"]
  codi = text["codi"]
  dades = obtenir_dades(codi)
  preu_base = dades["preu_base"]
  preu_final = calcular_preu_final(preu_base, edat)
  return {"nom":text["nom"], "edat": text["edat"], "codi": text["codi"], "categoria": dades["categoria"], "preu_base": preu_base, "fraude": dades["fraude"], "preu_final":preu_final }
```

### 4. ordenar_clients(clients)
- crea un diccionario clients_processats
- procesa los datos de cada cliente llamando la funcción processar_client
- almacena, uno por uno, los clientes procesados el diccionario creado
- ordena los elementos del diccionario en funcción de la probabilidad de fraude y el precio final, de mayor a menor
- devuelve el diccionario ordenado
  

```python
def ordenar_clients(clients):
 clients_processats = []
 for client in clients:
   client_processat = processar_client(client)
   clients_processats.append(client_processat)
 ranking = sorted(clients_processats, key=lambda c: (c['fraude'], c['preu_final']), reverse=True)
 return ranking
```

### generar_contrasenya()
- la funcción recibe como argumentos la longitud que debe tener la contrasenya y los tipos de caracteres a incluir en la contraseña: mayusculas, minusculas y numeros
- crea una lista (de caracteres) llamada contraseña
- elige al hazard una letra del alfabeto (ASCII), la transforma en mayuscula y la añade a la contraseña
- elige al hazard una letra del alfabeto (ASCII), la transforma en minuscula  y la añade a la contraseña
- elige al hazard un número y lo añade a la contraseña
- crea una "bolsa" intermedia de "caracteres de recambio" tipo string anomenada pool
- genera minusculas, mayusculas y numeros para añadir a la bolsa
- la funcción coge elementos de la bolsa al hazard y las añade a la contraseña pero solo mientras la longitud de la contraseña sea menor que la longitud total definida
- devuelve una contraseña con la longitud definida.

```python
import random
import string

def generar_contrasenya(longitud, majuscules=True, minuscules=True, numeros=True, signes=False):
  contrasenya = []
  if majuscules:
    contrasenya.append(random.choice(string.ascii_uppercase))
  if minuscules:
    contrasenya.append(random.choice(string.ascii_lowercase))
  if numeros:
    contrasenya.append(random.choice(string.digits))
  if signes:
    contrasenya.append(random.choice(string.punctuation))

  pool = ""
  if majuscules:
    pool += string.ascii_uppercase
  if minuscules:
    pool += string.ascii_lowercase
  if numeros:
    pool += string.digits
  if signes:
    pool += string.punctuation

  while len(contrasenya) < longitud:
    contrasenya.append(random.choice(pool))

  random.shuffle(contrasenya)

  contrasenya = "".join(contrasenya)

  return contrasenya

```

### 3. calcular_gols_equip()
- la funcción crea la variable equip para guardar cada equipo distincto de la columna local
- suma los puntos de cada equipo distincto en la columna gols-local (en casa)
- suma los puntos de cada equipo distincto en la columna gols-visitant (fuera de casa)
- suma los puntos en casa y fuera de casa de cada equipo
- suma los puntos de cada competidor en contra de un equipo dado que juega en su casa
- suma los puntos de cada competidor en contra de un equipo dado que juega fuera de su casa
- suma los puntos en casa y fuera de casa de cada competidor
- crea el dataframe taula con las nuevas columnas gols_favor y gols_contra
- cambia el nombre de la columna local a equip
- devuelve la taula con los goles marcados y recibidos por cada equipo

```python
def calcular_gols_equip(df):
  equip = df["local"].unique()
  gols_casa = df.groupby("local")["gols_local"].sum()
  gols_fora = df.groupby("visitant")["gols_visitant"].sum()
  gols_favor = gols_casa + gols_fora
  contra_casa = df.groupby("local")["gols_visitant"].sum()
  contra_fora = df.groupby("visitant")["gols_local"].sum()
  gols_contra = contra_casa + contra_fora
  taula = pd.DataFrame({"gols_favor": gols_favor, "gols_contra": gols_contra})
  taula.index.name = "equip"
  return taula
```

### 4. calcular_classificacio()

```python
def calcular_classificacio(df):

  df["victoria_local"] = df["gols_local"] > df["gols_visitant"]
  df["empat"] = df["gols_local"] == df["gols_visitant"]
  df["victoria_visitant"] = df["gols_local"] < df["gols_visitant"]
  gols_casa      = df.groupby("local")["gols_local"].sum()
  gols_fora = df.groupby("visitant")["gols_visitant"].sum()
  gols_favor = gols_casa + gols_fora
  contra_casa = df.groupby("local")["gols_visitant"].sum()
  contra_fora = df.groupby("visitant")["gols_local"].sum()
  gols_contra = contra_casa + contra_fora

  victories_casa = df.groupby("local")["victoria_local"].sum()
  victories_fora = df.groupby("visitant")["victoria_visitant"].sum()
  guanyats = victories_casa + victories_fora

  empats_casa = df.groupby("local")["empat"].sum()
  empats_fora = df.groupby("visitant")["empat"].sum()
  empatats = empats_casa + empats_fora

  derrotes_casa = df.groupby("local")["victoria_visitant"].sum()
  derrotes_fora = df.groupby("visitant")["victoria_local"].sum()
  perduts = derrotes_casa + derrotes_fora
  punts = guanyats*3 + empatats

  taula = pd.DataFrame({"gols_favor": gols_favor, "gols_contra": gols_contra, "guanyats": guanyats, "empatats": empatats, "perduts": perduts, "punts": punts})
  taula.index.name = "equip"
  taula = taula.sort_values("punts", ascending=False)
  return taula
```
